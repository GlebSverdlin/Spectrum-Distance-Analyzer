```python
mask_N2 = ((data['loglam']>math.log(5754, 10)) & (data['loglam']<math.log(5756, 10)))
mask_Ha = ((data['loglam']>math.log(6561, 10)) & (data['loglam']<math.log(6564, 10)))
mask_ZOI = ((data['loglam']>math.log(5000, 10)) & (data['loglam']<math.log(7000, 10)))
wave = 10**data['loglam']
flux = data['flux']
```

```python
plt.figure(figsize=(10,6))
plt.plot(wave[mask_ZOI], flux[mask_ZOI], 'b', alpha = 0.25)
plt.plot(wave[mask_N2], flux[mask_N2], 'red')
plt.plot(wave[mask_Ha], flux[mask_Ha], 'yellow')
```
